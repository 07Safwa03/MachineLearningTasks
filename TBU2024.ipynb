```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```

```python
# Hitung akurasi model
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)
print(f'Akurasi: {accuracy}')
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split

# Membuat dataset sintetis dengan 4 kelas
x, y = make_classification(n_samples=1775, n_features=3, n_informative=3, n_redundant=0,
                           n_clusters_per_class=1, n_classes=4, random_state=0)

# Membagi dataset menjadi training dan testing
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0)

# Membuat model Naive Bayes
model = GaussianNB()
model.fit(x_train, y_train)

# Membuat grid untuk plot decision boundary
x_min, x_max = x[:, 1].min() - 1, x[:, 1].max() + 1
y_min, y_max = x[:, 2].min() - 2, x[:, 2].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.1),
                     np.arange(y_min, y_max, 0.1))

# Menghitung prediksi untuk setiap titik dalam grid
Z = model.predict(np.c_[xx.ravel(), yy.ravel(), np.zeros_like(xx.ravel())])
Z = Z.reshape(xx.shape)

# Visualisasi decision boundary
plt.contourf(xx, yy, Z, alpha=0.8, cmap='coolwarm')

# Visualisasi data training dan testing
scatter = plt.scatter(x_test[:, 0], x_test[:, 1], c=y_test, marker='o', edgecolor='k', cmap='coolwarm')

# Menambahkan label dan judul
plt.xlabel('Fitur 1')
plt.ylabel('Fitur 2')
plt.title('Hasil Klasifikasi Naive Bayes dengan 4 Kelas tahun 2024')

# Menambahkan legenda
legend = plt.legend(*scatter.legend_elements(), title="Kelas")
plt.gca().add_artist(legend)

# Tampilkan plot
plt.show()
```
